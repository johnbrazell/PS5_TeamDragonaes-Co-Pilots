Run this notebook in the course idai610 kernel. Place the labeled competition CSVs beside the notebook, update the filenames below if needed, then choose **Run All**. This runs the frozen predictors with the course packages.

```python
from pathlib import Path

FRAUD_TEST = Path("fraud_test.csv")
ACTIVITY_TEST = Path("activity_test.csv")
OUTPUT = Path("results")
```

```python
from contextlib import redirect_stdout
from io import StringIO

from ps5_evaluate import run_evaluation

buffer = StringIO()
with redirect_stdout(buffer):
    run_evaluation(FRAUD_TEST, ACTIVITY_TEST, OUTPUT)
metrics = buffer.getvalue()
print(metrics, end="")
(OUTPUT / "metrics.txt").write_text(metrics, encoding="utf-8")
```
